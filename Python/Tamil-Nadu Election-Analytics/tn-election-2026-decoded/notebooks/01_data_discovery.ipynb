## Step 5 — Build the Winners Table (the foundation)

```python
# 5.1 Define MAJOR_PARTIES and the normalization function

# the parties we track explicitly. Everything else → 'OTHERS'.
MAJOR_PARTIES = {
    'DMK',         # Dravida Munnetra Kazhagam
    'AIADMK',      # All India Anna Dravida Munnetra Kazhagam
    'TVK',         # Tamilaga Vettri Kazhagam — new 2026 entrant
    'BJP',         # Bharatiya Janata Party
    'INC',         # Indian National Congress
    'PMK',         # Pattali Makkal Katchi
    'VCK',         # Viduthalai Chiruthaigal Katchi
    'NTK',         # Naam Tamilar Katchi
    'CPI',         # Communist Party of India
    'CPI(M)',      # Communist Party of India (Marxist)
    'DMDK',        # Desiya Murpokku Dravida Kazhagam
    'AMMK',        # Amma Makkal Munnetra Kazhagam
    'MDMK',        # Marumalarchi Dravida Munnetra Kazhagam
    'IND',         # Independent candidates
    'NOTA',        # None Of The Above
}

def normalize_party(p):
    """Bucket any party not in MAJOR_PARTIES into 'OTHERS'."""
    return p if p in MAJOR_PARTIES else 'OTHERS'

# Apply to both years — create a NEW column, don't overwrite raw 'party'
df_2021['party_clean'] = df_2021['party'].apply(normalize_party)
df_2026['party_clean'] = df_2026['party'].apply(normalize_party)

print("Normalization applied. Sample of new column:")
df_2026[['constituency','candidate','party','party_clean','votes']].head(10)
```

```python
# 5.2 Verify the normalization didn't break anything

print(f"=== 2021 party_clean counts ===")
print(df_2021['party_clean'].value_counts())
print(f"\n=== 2026 party_clean counts ===")
print(df_2026['party_clean'].value_counts())
```

```python
# 5.3 Build the winners table - the function

def build_winners_table(df, year_label):
    """
    Build one-row-per-AC table with:
      winner + runner-up info, margin, regional metadata.
    """
    # Sort: lowest ac_number first, then highest votes first.
    # That puts the winner at the top of each AC's group.
    s = df.sort_values(['ac_number', 'votes'], ascending=[True, False]).copy()

    # rank: 1 = winner, 2 = runner-up, 3+ = also-rans
    s['rank'] = s.groupby('ac_number').cumcount() + 1

    # Pull out winner rows
    winners = s[s['rank'] == 1][['ac_number', 'constituency', 'reserved', 'region',
                                   'candidate', 'party', 'party_clean', 'votes']].copy()
    winners = winners.rename(columns={
        'candidate':   'winner_candidate',
        'party':       'winner_party_raw',
        'party_clean': 'winner_party',
        'votes':       'winner_votes'
    })

    # Pull out runner-up rows
    runners = s[s['rank'] == 2][['ac_number', 'candidate', 'party_clean', 'votes']].copy()
    runners = runners.rename(columns={
        'candidate':   'runner_up_candidate',
        'party_clean': 'runner_up_party',
        'votes':       'runner_up_votes'
    })

    # Join winners + runners on ac_number
    out = winners.merge(runners, on='ac_number', how='left')

    # Total valid votes in the AC (denominator for margin %)
    totals = df.groupby('ac_number')['votes'].sum().rename('total_valid_votes').reset_index()
    out = out.merge(totals, on='ac_number')

    # Margin metrics
    out['margin']                  = out['winner_votes'] - out['runner_up_votes']
    out['margin_pct']              = (out['margin'] / out['total_valid_votes']) * 100
    out['winner_vote_share_pct']   = (out['winner_votes'] / out['total_valid_votes']) * 100

    # Add the year tag (useful when we concat the two tables later)
    out['year'] = year_label

    # Order columns nicely
    col_order = ['year', 'ac_number', 'constituency', 'region', 'reserved',
                 'winner_candidate', 'winner_party', 'winner_party_raw', 'winner_votes',
                 'runner_up_candidate', 'runner_up_party', 'runner_up_votes',
                 'margin', 'margin_pct', 'winner_vote_share_pct', 'total_valid_votes']
    return out[col_order]

winners_2021 = build_winners_table(df_2021, 2021)
winners_2026 = build_winners_table(df_2026, 2026)

print(f"winners_2021 shape: {winners_2021.shape}")
print(f"winners_2026 shape: {winners_2026.shape}")
```

```python
import plotly.graph_objects as go

# Build flow data from the winners tables
flow = (winners_2021[['ac_number','winner_party']]
        .merge(winners_2026[['ac_number','winner_party']],
               on='ac_number', suffixes=('_2021','_2026'))
        .groupby(['winner_party_2021','winner_party_2026'])
        .size()
        .reset_index(name='seats'))

# Order parties for visual coherence
party_order_2021 = ['DMK','AIADMK','INC','BJP','PMK','VCK','CPI(M)','CPI','OTHERS']
party_order_2026 = ['TVK','DMK','AIADMK','INC','BJP','PMK','VCK','CPI(M)','CPI','DMDK','AMMK','OTHERS']

# Sankey needs every node indexed by an integer
labels_2021 = [f"{p} (2021)" for p in party_order_2021 if p in flow['winner_party_2021'].values]
labels_2026 = [f"{p} (2026)" for p in party_order_2026 if p in flow['winner_party_2026'].values]
labels = labels_2021 + labels_2026
label_to_idx = {name: i for i, name in enumerate(labels)}

# Neutral, distinguishable color per node
color_map = {
    'TVK':    '#E63946',   # warm red (visually distinct, neutral)
    'DMK':    '#1D3557',   # navy
    'AIADMK': '#2A9D8F',   # teal
    'INC':    '#264653',   # dark slate
    'BJP':    '#F4A261',   # warm sand
    'PMK':    '#E9C46A',   # mustard
    'VCK':    '#6A4C93',   # purple
    'CPI':    '#8D99AE',
    'CPI(M)': '#A8DADC',
    'DMDK':   '#FFB703',
    'AMMK':   '#9D8189',
    'OTHERS': '#BDBDBD',   # grey
}
node_colors = [color_map.get(lbl.split(' (')[0], '#999') for lbl in labels]

# Sources, targets, values for the ribbons
sources = [label_to_idx[f"{r['winner_party_2021']} (2021)"] for _, r in flow.iterrows()]
targets = [label_to_idx[f"{r['winner_party_2026']} (2026)"] for _, r in flow.iterrows()]
values  = flow['seats'].tolist()

# Ribbon color = source party color, semi-transparent
def hex_to_rgba(h, alpha=0.4):
    h = h.lstrip('#'); return f"rgba({int(h[0:2],16)},{int(h[2:4],16)},{int(h[4:6],16)},{alpha})"
link_colors = [hex_to_rgba(color_map.get(labels[s].split(' (')[0], '#999')) for s in sources]

fig = go.Figure(go.Sankey(
    arrangement='snap',
    node=dict(
        pad=18, thickness=22,
        line=dict(color='black', width=0.5),
        label=labels, color=node_colors,
    ),
    link=dict(source=sources, target=targets, value=values, color=link_colors)
))

fig.update_layout(
    title_text="<b>Tamil Nadu Assembly: 2021 → 2026 Seat Flows</b><br>"
               "<span style='font-size:14px;color:#555'>234 constituencies. Ribbon width = seats moved between parties. Source: ECI.</span>",
    font=dict(size=12, family='Arial'),
    height=700, width=1100,
    paper_bgcolor='white',
)

# Save both PNG (for deck) and HTML (for dashboard)
CHARTS = REPO / "charts"
CHARTS.mkdir(exist_ok=True)
fig.write_html(CHARTS / "01_sankey_2021_to_2026.html")
try:
    fig.write_image(CHARTS / "01_sankey_2021_to_2026.png", scale=2)
    print("✅ Saved PNG + HTML")
except Exception as e:
    print(f"⚠️ HTML saved. PNG export failed (install kaleido: `pip install kaleido`): {e}")

fig.show()
```

## Chart 3 — Margin Buckets (the context beat)

```python
def margin_buckets(df):
    return {
        '>50% (decisive)':     int((df['winner_vote_share_pct'] > 50).sum()),
        '35–50% (plurality)':  int(((df['winner_vote_share_pct'] >= 35) & (df['winner_vote_share_pct'] <= 50)).sum()),
        '<35% (minority)':     int((df['winner_vote_share_pct'] < 35).sum()),
    }
b21 = margin_buckets(winners_2021)
b26 = margin_buckets(winners_2026)
cats = list(b21.keys())

fig3 = go.Figure()
fig3.add_trace(go.Bar(x=cats, y=list(b21.values()), name='2021',
                     marker_color='#94A3B8',
                     text=list(b21.values()), textposition='outside'))
fig3.add_trace(go.Bar(x=cats, y=list(b26.values()), name='2026',
                     marker_color='#E63946',
                     text=list(b26.values()), textposition='outside'))
fig3.update_layout(
    title='<b>How winners won: vote-share buckets</b><br>'
          '<span style="font-size:13px;color:#555">Number of constituencies in each bucket. Source: ECI.</span>',
    barmode='group',
    yaxis_title='Number of constituencies (of 234)',
    height=500, width=850,
    paper_bgcolor='white', plot_bgcolor='white',
    legend=dict(orientation='h', yanchor='bottom', y=1.02, xanchor='right', x=1)
)
fig3.update_yaxes(showgrid=True, gridcolor='#eee')

fig3.write_html(CHARTS / "03_margin_buckets.html")
fig3.show()
```

## Chart 4 — Regional Seats Side-by-Side (geographic context)

```python
from plotly.subplots import make_subplots

def region_seats(winners_df):
    seats = winners_df.groupby(['region','winner_party']).size().unstack(fill_value=0)
    keep = ['TVK','DMK','AIADMK','INC','BJP','PMK','VCK','DMDK','AMMK','OTHERS']
    seats = seats[[c for c in keep if c in seats.columns]]
    region_order = ['Chennai Metro','North','Central','Kongu','Delta','South']
    seats = seats.reindex(region_order)
    return seats

s21 = region_seats(winners_2021)
s26 = region_seats(winners_2026)

party_colors_full = {
    'TVK': '#E63946', 'DMK': '#1D3557', 'AIADMK': '#2A9D8F',
    'INC': '#264653', 'BJP': '#F4A261', 'PMK': '#E9C46A',
    'VCK': '#6A4C93', 'DMDK': '#FFB703', 'AMMK': '#9D8189', 'OTHERS': '#BDBDBD',
}

fig4 = make_subplots(rows=1, cols=2, subplot_titles=('<b>2021</b>', '<b>2026</b>'), shared_yaxes=True)

seen = set()
for party in s21.columns:
    fig4.add_trace(go.Bar(name=party, x=s21.index, y=s21[party],
                          marker_color=party_colors_full.get(party,'#888'),
                          legendgroup=party, showlegend=(party not in seen)), row=1, col=1)
    seen.add(party)
for party in s26.columns:
    fig4.add_trace(go.Bar(name=party, x=s26.index, y=s26[party],
                          marker_color=party_colors_full.get(party,'#888'),
                          legendgroup=party, showlegend=(party not in seen)), row=1, col=2)
    seen.add(party)

fig4.update_layout(
    title='<b>Seats won by region — 2021 vs 2026</b><br>'
          '<span style="font-size:13px;color:#555">6 editorial regions; 234 seats total. Source: ECI.</span>',
    barmode='stack',
    height=550, width=1150,
    paper_bgcolor='white', plot_bgcolor='white',
    legend=dict(orientation='h', yanchor='bottom', y=-0.18, xanchor='center', x=0.5)
)
fig4.update_yaxes(title_text='Seats', row=1, col=1)
fig4.update_xaxes(tickangle=-30)

fig4.write_html(CHARTS / "04_regional_seats.html")
fig4.show()
```
